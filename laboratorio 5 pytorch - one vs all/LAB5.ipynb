Despues de la verificamos cargamos el modelo

```python
data = pd.read_csv('personalidad.csv', delimiter=',')

# Mapear y a valores enteros
mapping = {'Introvert': 0, 'Extrovert': 1, 'Ambivert': 2}
data["personalidad"] = data["personality_type"].map(mapping)
X = data.drop(columns=["personality_type", "personalidad"])  # todas menos las columnas personality_type y personalidad
Y = data["personalidad"]
data
```

Verificamos los valores de X Y

```python
print("Características de entrada (X):")
print(X)
print("Etiquetas de salida (y):")
print(Y)
```

```python
x_2=np.array(X)
y_2=np.array(Y)

# Normalización y split
X_train = x_2[:16000]  # primeros 16k para entrenamiento
X_test  = x_2[16000:]  # últimos 4k para prueba

# Normalización 0-1
X_train = (X_train - 1) / 9
X_test  = (X_test - 1) / 9

# Convertir etiquetas a enteros
y_train = y_2[:16000].astype(np.int32)
y_test  = y_2[16000:].astype(np.int32)
```

Mide qué tan bien el modelo predijo la clase verdadera.

Si la probabilidad de la clase correcta es alta → pérdida baja.

Si es baja → pérdida alta.

```python
# función de pérdida y derivada

def softmax(x):
    return torch.exp(x) / torch.exp(x).sum(axis=-1,keepdims=True)

def cross_entropy(output, target):
    logits = output[torch.arange(len(output)), target]
    loss = - logits + torch.log(torch.sum(torch.exp(output), axis=-1))
    loss = loss.mean()
    return loss
```

Convierte tus datos de entrada (X_train, que es un numpy.ndarray) en un tensor de PyTorch y mueve ese tensor a la GPU, para que los cálculos se hagan más rápido que en CPU.

```python
X_t = torch.tensor(X_train, dtype=torch.float32).cuda()
Y_t = torch.tensor(y_train, dtype=torch.long).cuda()
model = model.cuda()
```

# bucle entrenamiento
epochs         umero de iteraciones

lr             pasos de aprendizaje

log_each       cada cuántas iteraciones mostramos el log

l = []         lista donde guardamos los valores de la función de pérdida

```python
epochs = 350 
lr = 0.8  
log_each = 10 
l = []
for e in range(1, epochs + 1):

    # forward
    y_pred = model(X_t)

    # loss
    loss = cross_entropy(y_pred, Y_t)
    l.append(loss.item())

    # ponemos a cero los gradientes
    model.zero_grad()

    # Backprop (calculamos todos los gradientes automáticamente)
    loss.backward()

    # update de los pesos
    with torch.no_grad():
        for param in model.parameters():
            param -= lr * param.grad

    if not e % log_each:
        print(f"Epoch {e}/{epochs} Loss {np.mean(l):.5f}")
```

Explicacion del paso a paso

.eval: cambia al modo evaluacion

softmax(): devuelve las probabilidades entre 0 y 1

torch.argmax: la clase con la mayor probabilidad

accuracy_score(x,y): compara ambas listas y devuelve la probabilidad
la prediccion de la red estara en valores entre 0 y 1, se lo multiplica *100 y se tendra la probabilidad exacta

```python
from sklearn.metrics import accuracy_score

def evaluate(x):
    model.eval()
    y_pred = model(x)  # predecir
    # calcular las clases predichas
    y_probas = softmax(y_pred)  # probabilidades entre 0 y 1
    return torch.argmax(y_probas, axis=1)  # clase con mayor probabilidad

y_pred = evaluate(torch.from_numpy(X_test).float().cuda())
accuracy_score(y_test, y_pred.cpu().numpy())
```

en este caso el entrenamiento fu exitoso, los valores estan variando entre 0.9962 y 0.997 lo cual el margen de error es menor al 1%
